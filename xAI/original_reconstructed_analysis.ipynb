```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat

# =========================
# Config
# =========================
analysis_type = "N400"  # e.g., "MMN", "N170", "N400", "P3"
datain_dir = os.path.join(os.getcwd(), "datain")

ori_path = os.path.join(datain_dir, f"{analysis_type}_original_betas.mat")
rec_path = os.path.join(datain_dir, f"{analysis_type}_reconstructed_betas.mat")

# Choose normalization method per subject:
# - "max": divide by subject-wise max over channels (recommended)
# - "sum": divide by subject-wise sum over channels
# - "zscore": z-score across channels for each subject (keeps sign; less common for MAE)
norm_method = "max"

# If you have channel names, put them here; otherwise will use 1..nChan
channel_name = None  # e.g., {0:"FP1", 1:"F3", ...}  (optional)

# =========================
# Load data
# =========================
ori_betas_mat = loadmat(ori_path, squeeze_me=True, struct_as_record=False)
rec_betas_mat = loadmat(rec_path, squeeze_me=True, struct_as_record=False)

# Adjust keys if needed
ori_betas = np.array(ori_betas_mat["data"], dtype=float)          # [nBeta, nTime, nChan, nSubject]
learned_betas = np.array(rec_betas_mat["recon_betas"], dtype=float)

print("Original betas shape:", ori_betas.shape)
print("Learned betas shape:", learned_betas.shape)

if ori_betas.shape != learned_betas.shape:
    raise ValueError(f"Shape mismatch: ori {ori_betas.shape} vs recon {learned_betas.shape}")

nBeta, nTime, nChan, nSubj = ori_betas.shape

# =========================
# 1) Per-subject channel MAE
# =========================
# abs error: [nBeta, nTime, nChan, nSubject]
abs_err = np.abs(learned_betas - ori_betas)

# Channel MAE per subject:
# mean over beta and time -> [nChan, nSubject]
chan_mae_subj = abs_err.mean(axis=(0, 1))  # average across (beta,time)

# Optional: if you want to average over beta only, keeping time, use:
# chan_time_mae_subj = abs_err.mean(axis=0)  # [nTime, nChan, nSubject]

print("chan_mae_subj shape [nChan, nSubject]:", chan_mae_subj.shape)

# =========================
# 2) Normalize per subject
# =========================
eps = 1e-12
if norm_method == "max":
    denom = np.maximum(chan_mae_subj.max(axis=0, keepdims=True), eps)   # [1, nSubj]
    chan_mae_norm = chan_mae_subj / denom
elif norm_method == "sum":
    denom = np.maximum(chan_mae_subj.sum(axis=0, keepdims=True), eps)
    chan_mae_norm = chan_mae_subj / denom
elif norm_method == "zscore":
    mu = chan_mae_subj.mean(axis=0, keepdims=True)
    sd = np.maximum(chan_mae_subj.std(axis=0, ddof=1, keepdims=True), eps)
    chan_mae_norm = (chan_mae_subj - mu) / sd
else:
    raise ValueError("norm_method must be one of: 'max', 'sum', 'zscore'")

# =========================
# 3) Aggregate across subjects
# =========================
# Mean across subjects -> [nChan]
agg_chan_mae = chan_mae_norm.mean(axis=1)

# Sort channels by aggregated MAE (optional)
order = np.argsort(agg_chan_mae)[::-1]
agg_sorted = agg_chan_mae[order]

# Labels
if isinstance(channel_name, dict) and len(channel_name) == nChan:
    labels = [channel_name[i] for i in range(nChan)]
else:
    labels = [str(i + 1) for i in range(nChan)]  # 1..nChan

labels_sorted = [labels[i] for i in order]

# =========================
# Plot
# =========================
plt.figure(figsize=(12, 5))
x = np.arange(nChan)
plt.bar(x, agg_sorted)
plt.xticks(x, labels_sorted, rotation=90, fontsize=10)
plt.ylabel("Aggregated (subject-normalized) channel MAE", fontsize=12)
plt.xlabel("Channel (sorted by aggregated MAE)", fontsize=12)
plt.title(f"{analysis_type}: Aggregated Channel Reconstruction MAE (norm={norm_method})", fontsize=13)
plt.tight_layout()
plt.show()

# If you also want the raw (non-normalized) aggregated MAE across subjects:
raw_agg_chan_mae = chan_mae_subj.mean(axis=1)  # [nChan]
print("Top-5 channels by normalized aggregated MAE:", [(labels_sorted[i], float(agg_sorted[i])) for i in range(min(5, nChan))])
print("Top-5 channels by raw aggregated MAE:", [(labels[i], float(raw_agg_chan_mae[i])) for i in np.argsort(raw_agg_chan_mae)[::-1][:5]])
```

```python
# shape [nBeta, nSubject, nChan, nTime]
ori_betas = np.transpose(ori_betas, (0, 3, 2, 1))
learned_betas = np.transpose(learned_betas, (0, 3, 2, 1))
print("Original betas shape:", ori_betas.shape)
print("Learned betas shape:", learned_betas.shape)
mae = np.abs((ori_betas - learned_betas) ** 2)
nBeta, nSubj, nChan, nTime = mae.shape
print("MAE mean:", np.mean(mae))
yticks = np.arange(1, nChan+1)
time = np.arange(nTime)
chan_time_mae = mae.mean(axis=(0,1))
```

```python
# heat-map of all beta and subject

plt.figure(figsize=(16,13))
plt.imshow(chan_time_mae, aspect='auto', origin='lower', cmap='Reds')
plt.colorbar(label='MAE')
plt.yticks(np.arange(nChan), yticks)
plt.xlabel('Time')
plt.ylabel('Channel')
plt.title(analysis_type+'Channel_Time MAE')
plt.tight_layout()
plt.savefig(os.path.join(save_dir, analysis_type+'_Channel_Time MAE.png'))
#plt.show()
plt.close()
```

```python
save_sub_dir1 = os.path.join(save_dir, 'Each subject MAE')
os.makedirs(save_sub_dir1, exist_ok=True)

global_max = np.percentile(mae, 1)
global_min = np.percentile(mae, 99)

# heat-map per subject
for iSubj in range(nSubj):
    chan_time_mae = mae[:, iSubj].mean(axis=0)
    plt.figure(figsize=(16,13))
    plt.imshow(chan_time_mae, aspect='auto', origin='lower', cmap='Reds', vmin=global_min, vmax=global_max)
    plt.colorbar(label='MAE')
    plt.yticks(np.arange(nChan), yticks)
    plt.xlabel('Time')
    plt.ylabel('Channel')
    plt.title(f'{analysis_type}_Subject_{iSubj}_Channel_Time MAE')
    plt.tight_layout()
    plt.savefig(os.path.join(save_sub_dir1, f'Subject_{iSubj}_Channel_Time_MAE.png'))
    #plt.show()
    plt.close()
```

```python
save_sub_dir2 = os.path.join(save_dir, 'Each beta MAE')
os.makedirs(save_sub_dir2, exist_ok=True)

# heat-map per beta
for iBeta in range(nBeta):
    chan_time_mae = mae[iBeta].mean(axis=0)
    plt.figure(figsize=(16,13))
    plt.imshow(chan_time_mae, aspect='auto', origin='lower', cmap='Reds', vmin=global_min, vmax=global_max)
    plt.colorbar(label='MAE')
    plt.yticks(np.arange(nChan), yticks)
    plt.xlabel('Time')
    plt.ylabel('Channel')
    plt.title(f'{analysis_type}_Beta_{iBeta}_Channel * Time MAE')
    plt.tight_layout()
    plt.savefig(os.path.join(save_sub_dir2, f'Beta_{iBeta}_Channel_Time_MAE.png'))
    #plt.show()
    plt.close()
```

```python
print("ori_betas mean:", np.mean(ori_betas))
print("learned_betas mean:", np.mean(learned_betas))
save_sub_dir3 = os.path.join(save_dir, 'Reconstruction vs original value comparison')
os.makedirs(save_sub_dir3, exist_ok=True)

# reconstruction and original value comparison per iBeta
for iBeta in range(nBeta):
    ori_mean = ori_betas[iBeta].mean(axis=(0, 1))
    recon_mean = learned_betas[iBeta].mean(axis=(0, 1))

    plt.figure(figsize=(16,13))
    plt.plot(time, ori_mean, label='original', color='blue')
    plt.xlabel('Time')
    plt.ylabel('Amplitude')
    plt.title(f'{analysis_type} | Beta {iBeta+1}')
    plt.tight_layout()
    plt.savefig(os.path.join(save_sub_dir3, f'Beta{iBeta+1}_original_TimeSeries.png'))
    plt.close()
    
    plt.figure(figsize=(16,13))
    plt.plot(time, recon_mean, label='reconstructed', color='red')
    plt.xlabel('Time')
    plt.ylabel('Amplitude')
    plt.title(f'{analysis_type} | Beta {iBeta+1}')
    plt.tight_layout()
    plt.savefig(os.path.join(save_sub_dir3, f'Beta{iBeta+1}_reconstructed_TimeSeries.png'))
    plt.close()
```

```python
# reconstruction and original value comparison per iBeta and channel

for iBeta in range(nBeta):
    save_sub_dir4 = os.path.join(save_sub_dir3, f'Beta_{iBeta+1}_per_channel')
    os.makedirs(save_sub_dir4, exist_ok=True)
    for iChan in range(nChan):
        ori_mean = ori_betas[iBeta,:,iChan,:].mean(axis=0)
        recon_mean = learned_betas[iBeta,:,iChan,:].mean(axis=0)
        plt.figure(figsize=(16,13))
        plt.plot(time, ori_mean, label='original', color='blue')
        plt.plot(time, recon_mean, label='reconstructed', color='red')
        plt.xlabel('Time')
        plt.ylabel('Amplitude')
        plt.title(f'Beta {iBeta+1} | Channel {iChan+1}')
        plt.tight_layout()
        plt.legend()
        plt.savefig(os.path.join(save_sub_dir4, f'Beta{iBeta+1}_Channel{iChan+1}_TimeSeries.png'))
        plt.close()
```

```python
# Top 5 channels with highest MAE per beta
print("top5 highest MAE channels per beta:")

for iBeta in range(nBeta):
    chan_mae = mae[iBeta].mean(axis=(0, 2))   # (nChan,)
    top5_idx = np.flip(np.argsort(chan_mae)[-5:])
    top5_vals = chan_mae[top5_idx]
    
    
    print(f"\n========== Beta {iBeta+1} ==========")
    for i in range(len(top5_idx)):
        print(f"#{i+1}: Channel {top5_idx[i]+1:<2}  MAE={top5_vals[i]:.6f}")
```
