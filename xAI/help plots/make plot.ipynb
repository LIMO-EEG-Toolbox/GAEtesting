```python
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np

def plot_block(ax, img_path, text="", duration=""):
    ax.set_facecolor((0.7, 0.7, 0.7))
    ax.set_xticks([])
    ax.set_yticks([])
    
    if img_path:
        img = Image.open(img_path)
        ax.imshow(img)
    else:
        ax.text(0.5, 0.5, text, fontsize=26, ha='center', va='center')
    
    ax.text(0.5, -0.05, duration, fontsize=14, ha='center', va='top', transform=ax.transAxes)

fig = plt.figure(figsize=(12, 6))

stimuli = [
    (r"E:\MThesis\GAEtesting\help plots\N170.jpg",     "300 ms"),
    (None,          "1100-1300 ms"),
    (r"E:\MThesis\GAEtesting\help plots\N170.jpg",    "300 ms"),
    (None,          "1100-1300 ms"),
    ("E:\MThesis\GAEtesting\help plots\N170.jpg",   "300 ms"),
]

for i, (img, dur) in enumerate(stimuli):
    ax = fig.add_subplot(1, len(stimuli), i+1)
    plot_block(ax, img_path=img, text="Fix" if img is None else "", duration=dur)

plt.suptitle("Example: Face Perception N170", fontsize=18)
plt.tight_layout()
plt.show()
```
